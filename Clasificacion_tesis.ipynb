**Convercion de datos**

```python
from sklearn.preprocessing import TargetEncoder
from sklearn.model_selection import train_test_split

X = df[['alcaldia', 'colonia']]
y = df['indice_des']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

encoder = TargetEncoder(
    target_type='multiclass',
    cv=5,
    smooth='auto',
    random_state=42
)

X_train_enc = encoder.fit_transform(X_train, y_train)
X_test_enc = encoder.transform(X_test)
```

```python
#Convertir los x_train_enc en dataframe

X_train_enc = pd.DataFrame(
    X_train_enc,
    columns=encoder.get_feature_names_out(),
    index=X_train.index
)
```

```python
#Convertir los x_test_enc en dataframe

X_test_enc = pd.DataFrame(
    X_test_enc,
    columns=encoder.get_feature_names_out(),
    index=X_test.index
)
```

```python
#Unir los datasets x_train_enc y x_test_enc

X_encoded = pd.concat([X_train_enc, X_test_enc])
```

```python
#Ordenar las instancias X_encoded

X_encoded = X_encoded.sort_index()
```

```python
#Se crea una copia de df para crear una version de un dataset con k-fol Target Encoding
df_KFTE = df.copy()
```

```python
#Se eliminan las columnas alcaldia y colonia
df_KFTE = df_KFTE.drop(columns=['alcaldia', 'colonia'])
```

```python
#Se agrgan las nuevas columnas a df_KFTE

df_KFTE = pd.concat([df_KFTE, X_encoded], axis=1)
```
